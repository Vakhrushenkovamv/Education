# 1.Подготовка датасета

**Импорт библиотек**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# разбиение на train и test
from sklearn.model_selection import train_test_split
```

**Загрузка данных**

```python
csv= pd.read_csv('flavors_of_cacao.csv')
```

**Выведем первые 5 строк, посмотрим на структуру данных**

```python
csv.head(5)
```

**Проверяем типы данных**

```python
csv.info()
```

**Заменим переносы строк пробелом для визуального удобстава**

```python
csv.columns = (
    csv.columns
       .str.replace(r'\s+', ' ', regex=True)
       .str.strip()
)
```

```python
csv.info()
```

**Посмотрим на размер датасета**

```python
csv.shape
```

*У нас есть пропуски в признаках Bean Type и Broad Bean Origin- на ону строку меньше, чем остальные значения*

**Опишем суть нашего датасета**

<span style="color: darkviolet;">**Описание датасета:**</span>  
Датасет содержит рейтинг шоколадных батончиков.  
набор данных содержит экспертные оценки более 1700 отдельных шоколадных батончиков, а также информацию об их региональном происхождении, процентном содержании какао, сорте используемых какао-бобов и месте их выращивания.  
<span style="color: darkviolet;">**Описание признаков:**</span>  
- **Company (Maker-if known)**- Название компании — производителя шоколада
- **Specific Bean Origin or Bar Name**- Конкретное место происхождения какао-бобов или название шоколадного батончика/линейки
- **REF**- Идентификатор записи, связанный с моментом добавления отзыва в базу: чем выше значение, тем позднее была добавлена запись
- **Review Date**- Год публикации обзора
- **Cocoa Percent**- Процентное содержание какао в шоколаде
- **Company Location**- Страна расположения компании-производителя
- **Rating**- Рейтинг. Оценка качества шоколада
- **Bean Type**- Сорт или тип какао-бобов
- **Broad Bean Origin**- Более широкая географическая область происхождения какао-бобов: страна или регион

Целевая переменная **Rating** Будем предсказывать рейтинг шоколадных батончиков.

**Проверяем количество пропущенных значений**

```python
csv.isnull().sum() 
```

Есть по одному пропущенному значению в признаке **Bean Type** :Сорт или тип какао-бобов и  
**Broad Bean Origin**: географическая область происхождения какао-бобов: страна или регион

Так как пропусков мало наиболее правильный вариант их удалить чтобы очистить наши данные

```python
csv.dropna(axis=0 ,subset=['Bean Type', 'Broad Bean Origin'], inplace=True)
```

**Проверяем остались ли пропуски в наших столбцах**

```python
csv.isnull().sum() 
```

**Пропусков нет**

В нашем датасете в признаке **Cocoa Percent** тип данных **str**, однако речь идет о процентном содержании какао в шоколаде. Данный признак лучше преобразовать в числовой тип, предварительно очистив знак процента в строках

**Очищаем знак процента в строках**

```python
csv['Cocoa Percent'] = csv['Cocoa Percent'].str.strip('%')
```

**Преобразуем тип данных в float**

```python
csv['Cocoa Percent']= csv['Cocoa Percent'].astype('float')
```

```python
csv.info()
```

**Так же правильнее удалить признак REF, так как это техническое поле**

```python
csv = csv.drop(columns=['REF'])
```

```python
csv.info()
```

*Проведем работу с неоднозначным категориальным признаком Specific Bean Origin or Bar Name. По наименованию признака можно судить, что в нем намешаны разные по смыслу значения. Посмотрим кол-во уникальных значений*

```python
csv['Specific Bean Origin or Bar Name'].nunique()
```

*Уникальных значений больше 50% к тому же разные по смыслу значения в категории, которые могут встречаться это как и происхождение какао-бобов и коммерческое название продукта. Принимаю решение удалить такой столбец, чтобы не путать модель.* 

```python
csv = csv.drop(columns=['Specific Bean Origin or Bar Name'])
```

```python
csv.info()
```

*В данных есть текстовые (категориальные) колонки, преобразуем их в числа:*


```python
# Сначала в переменную поместим наши категориальные признаки:  
cat_features = [
    'Company (Maker-if known)',
    'Company Location',
    'Bean Type',
    'Broad Bean Origin'
]
```

```python
# кодируем
csv_encoded = pd.get_dummies(
    csv,
    columns=cat_features,
    dtype=int
)
```

```python
csv_encoded.head()
```

**Отделяем целевую переменную и признаки**

```python

X = csv_encoded.drop(columns=['Rating'])
y = csv_encoded['Rating']
```

**Разделяем данные на обучающие и тестовые**

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42
)
```

```python
print(X_train.shape, y_train.shape)
```

```python
print(X_test.shape,  y_test.shape)
```

## 2.Обучаем нашу модель

```python
# Выбираем линейную модель. Линейную регрессию для числового прогноза
# Импортируем LinearRegression
from sklearn.linear_model import LinearRegression 

# Создаем объект линейной регрессии
model = LinearRegression() 

# Обучаем модель

model.fit(X_train, y_train) 

```

# 3.Получаем предсказания модели 

```python
y_pred = model.predict(X_test)
```

```python
print(y_pred[:5])
```

# 4.Считаем метрики

```python
# Импортируем подходящую метрику

from sklearn.metrics import mean_absolute_error

print(mean_absolute_error(y_test, y_pred))
```

```python
print("Настоящие значения:")
print(y_test[:15])

print("Предсказанные значения:")
print(y_pred[:15])
```

**Посчитаем коэфициент детерминации**

```python
from sklearn.metrics import r2_score
print('R2:', np.round(r2_score(y_test, y_pred),2))
```

**Метрика выдала показания хуже прогноза среднего. Возможно слишком много признаков после кодирования методом get_dummies или слабая линейная зависимость между признаками и рейтингами**

*Попробуем обучить модель с помощью дерева решений (DecisionTreeRegressor)*

```python
from sklearn.tree import DecisionTreeRegressor
# Попробуем ограничить глубину дерева до 5, а так же минимальное кол-во объектов до 5)
model = DecisionTreeRegressor(
    max_depth=5,
    min_samples_leaf=5,
    random_state=42)
# Обучаем
model.fit(X_train, y_train)
```

**Получаем предсказания модели на тестовой выборке**

```python
y_pred = model.predict(X_test)
```

```python
print(y_pred[:10])
```

```python
print('R2:',r2_score(y_test, y_pred))
```

```python
print(mean_absolute_error(y_test, y_pred))
```

*По сравнению с предыдущим результатом метрика показала более лучший результат. Модель немного лучше постоянного прогноза, равного среднему значению y_test, однако предсказательная способность модели пока остаётся слабой. На мой взгляд надо поработать более тщательно с датасетом, а именно с высокой кардинальностью признаков и шифрованием категориальных признаков, возможно с выявлением зависимостей целевой переменной от признаков.* 

```python
print('Train R²:', r2_score(y_train, model.predict(X_train)))
print('Test R²:', r2_score(y_test, model.predict(X_test)))
```

*На обучающей выборке коэффициент детерминации составил 0.207, а на тестовой — 0.025. Таким образом, модель объясняет около 20.7% вариации рейтинга на обучающих данных и только 2.5% на новых данных. Снижение показателя на тестовой выборке указывает на частичное переобучение. При этом невысокое значение R² даже на обучающей выборке свидетельствует о слабой зависимости рейтинга от используемых признаков или о недостаточной способности модели выявить эту зависимость.*
